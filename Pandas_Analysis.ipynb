```python
import pandas as pd
import numpy as np

data = {
    'transaction_id': range(1, 21),
    'date': pd.date_range('2024-10-01', periods=20, freq='D'),
    'region': ['North', 'South', 'East', 'West', None, 'North', 'South', None, 'East', 'West',
               'North', 'South', 'East', 'West', 'North', None, 'East', 'West', 'North', 'South'],
    'product_category': ['Electronics', 'Clothing', None, 'Books', 'Electronics', 'Home',
                         'Clothing', 'Books', 'Electronics', None, 'Home', 'Clothing',
                         'Books', 'Electronics', 'Home', 'Clothing', 'Books', 'Electronics',
                         'Home', 'Clothing'],
    'sales_amount': [1200, 450, 890, None, 1500, 670, None, 340, 2100, 780,
                     560, None, 420, 1800, 920, 510, 380, None, 1100, 640],
    'quantity': [2, 5, 3, 1, None, 4, 2, 3, 1, 5,
                 3, 2, None, 1, 4, 3, 2, 1, None, 4],
    'customer_age': [25, 34, None, 45, 29, None, 38, 52, 27, 41,
                     33, None, 48, 26, 35, 42, None, 31, 39, 44],
    'payment_method': ['Credit Card', 'UPI', 'Cash', 'Debit Card', 'Credit Card',
                       'UPI', 'Cash', None, 'Credit Card', 'UPI', 'Debit Card',
                       'Cash', 'Credit Card', None, 'UPI', 'Cash', 'Debit Card',
                       'Credit Card', 'UPI', None]
}

df = pd.DataFrame(data)
df
```

Task 2: Handle Missing Values
Region & Product_category: Fill with mode (most frequent value)

```python
df['region'] = df['region'].fillna(df['region'].mode()[0])
df['product_category'] = df['product_category'].fillna(df['product_category'].mode()[0])
df['region']
```

```python
df['product_category']
```

Sales_amount: Fill with median

```python
median_value = df['sales_amount'].median()
median_value
df['sales_amount'] = df['sales_amount'].fillna(median_value)
print(df['sales_amount'])
print("Missing values after filling:")
print(df['sales_amount'].isna().sum())
```

Quantity: Fill using forward fill (ffill)

```python
df['quantity'] = df['quantity'].fillna(method='ffill')
print(df['quantity'])
print(df['quantity'].isna().sum())
```

Customer_age: Fill with mean (rounded to integer)

```python
age_mean = round(df['customer_age'].mean())
print(age_mean)
df['customer_age'] = df['customer_age'].fillna(mean_age)
print(df['customer_age'])
print(df['customer_age'].isna().sum())
```

```python
df['payment_method'].isna()
```

Payment_method: Drop rows where payment_method is missing

```python
df = df.dropna(subset=['payment_method'])
df
```

```python
print("Missing values in payment_method:")
print(df['payment_method'].isna().sum())
```

```python
print("\nUpdated DataFrame shape:")
print(df.shape)
```

Task 3: GroupBy Analysis
Calculate total sales by region

```python
region_sales = df.groupby('region')['sales_amount'].sum()

print(region_sales)
```

Calculate average sales by product_category

```python
category_avg_sales = df.groupby('product_category')['sales_amount'].mean()

print(category_avg_sales)
```

Group by both region and product_category, calculate total sales and quantity, then use reset_index() to flatten

```python
grouped_df = df.groupby(['region', 'product_category']).agg({
    'sales_amount': 'sum',
    'quantity': 'sum'
}).reset_index()
print(grouped_df)
```

Display top 3 region-product combinations by sales

```python
top3 = grouped_df.sort_values(by='sales_amount', ascending=False).head(3)
top3
```

Task 4: Custom Aggregation: Create a function sales_range() that returns max - min of sales

```python
def sales_range(series):
    return series.max() - series.min()

range_value = sales_range(df['sales_amount'])

print("Sales Range (Max - Min):", range_value)
```

Apply it to find sales range for each region

```python
region_sales_range = df.groupby('region')['sales_amount'].apply(sales_range).reset_index()
region_sales_range
```

Use .agg() to calculate multiple metrics by region:
sales_amount: sum, mean, max
quantity: sum, min

```python
region_summary = df.groupby('region').agg({
    'sales_amount': ['sum', 'mean', 'max'],
    'quantity': ['sum', 'min']
}).reset_index()

region_summary
```
