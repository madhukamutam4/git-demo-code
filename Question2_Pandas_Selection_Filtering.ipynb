```python
import pandas as pd
print("*" * 50)
print("Task 1 — Create DataFrame")
print("*" * 50)

data = {
    "EmployeeID": [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],
    "Name": ["Sarah", "Michael", "Jennifer", "David", "Lisa",
             "James", "Emma", "Robert", "Sophia", "William"],
    "Department": ["Engineering", "Sales", "Marketing", "Engineering", "HR",
                   "Sales", "Engineering", "Marketing", "Sales", "HR"],
    "Salary": [85000, 62000, 71000, 92000, 58000,
               68000, 78000, 75000, 65000, 60000],
    "Performance Score": [4, 3, 5, 5, 4,
                          4, 3, 4, 5, 3],
    "Years of Service": [5, 3, 7, 8, 4,
                         6, 2, 5, 4, 3]
}

df = pd.DataFrame(data)
print(df)

print("\n" + "*" * 50)
print("Task 2 — Select Name, Department, Salary")
print("*" * 50)

print(df[['Name', 'Department', 'Salary']])

print("\n" + "*" * 50)
print("Task 3 — Use iloc (Row Position 5)")
print("*" * 50)

print(df.iloc[5])
```

```python
print("\n" + "*" * 50)
print("Task 4 — Use loc (EmployeeID 208)")
print("*" * 50)

df.set_index("EmployeeID", inplace=True)
print(df.loc[208])
```

```python
print("\n" + "*" * 50)
print("Task 5 — Employees from Engineering")
print("*" * 50)

print(df[df["Department"] == "Engineering"])
```

```python
print("\n" + "*" * 50)
print("Task 6 — Salary > 70000")
print("*" * 50)

print(df[df["Salary"] > 70000])
```

```python
print("\n" + "*" * 50)
print("Task 7 — Sales with Performance Score >= 4")
print("*" * 50)

print(df[(df["Department"] == "Sales") &
         (df["Performance Score"] >= 4)])
```

```python
print("\n" + "*" * 50)
print("Task 8 — Top 3 Highest Paid Employees")
print("*" * 50)

sorted_salary = df.sort_values(by="Salary", ascending=False)
print(sorted_salary.head(3))
```

```python
print("\n" + "*" * 50)
print("Task 9 — Sort by Department (ASC) & Salary (DESC)")
print("*" * 50)

sorted_dept_salary = df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
)

print(sorted_dept_salary)
```
